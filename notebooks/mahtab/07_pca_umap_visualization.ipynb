# Raisin Detective – PCA and UMAP Visualization

In this notebook, we use PCA and UMAP to visualize the Raisin dataset in two dimensions.

The goal is not to build the main classification model, but to explore whether the two raisin classes, Kecimen and Besni, form visible patterns or clusters when the numerical features are reduced to two dimensions.

PCA is a linear dimensionality reduction method, while UMAP is a non-linear method that can capture more complex structures in the data.
